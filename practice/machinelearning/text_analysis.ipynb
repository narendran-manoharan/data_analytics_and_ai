```python
def plot_wordcloud(wordcloud):
    plt.figure(figsize=(10,6))
    plt.axis("off")
    plt.imshow(wordcloud)
    plt.show()
```

```python
from wordcloud import WordCloud, STOPWORDS

wordcloud = WordCloud(
    width = 800, 
    height = 400,
    background_color = "salmon",
    colormap = "Pastel1",
    collocations = False, 
    stopwords = STOPWORDS, 
).generate(text)

plot_wordcloud(wordcloud)
```

```python
wordcloud.to_file("wordcloud.png")
```
